```python
class Book():

    def __init__(self, title, author, pages):

        self.title = title 
        self.author = author
        self.pages = pages

    def __str__(self): #Returns back the actual string representation 
        return f"{self.title} by {self.author}"#so you can print out user defined objects


    def __len__(self): #Returns back the length so you can return back the length of user-defined objects
        return self.pages

    def __del__(self):
        print("A book object has been deleted")
```

```python
b = Book('Python rocks', 'Jose', 200)
```

```python
print(b)
```

```python
str(b)
```

```python
len(b)
```

```python
del b
```
