```python
class Circle():
    #CLASS OBJECT ATRIBUTE
    pi = 3.14

    def __init__(self, radius=1):
        self.radius = radius
        self.area = radius*radius*self.pi #Attribute for area

    #METHOD
    def get_circumference(self): #self to represent a particular instance of an object 
        return self.radius * Circle.pi *2 #Because pi is a class object attribute, you can call it both ways
```

```python
my_circle = Circle()
```

```python
#INHERITENCE AND POLYMORPHISM:
#INHERITANCE: A WAY TO FORM NEW CLASSES USING CLASSES THAT ARE ALREADY DEFINED 
```

```python
class Animal(): #Base Class
    def __init__(self):
        print("Animal Created")

    def who_am_i(self):
        print("I am an animal")

    def eat(self):
        print("I am eating")
```

```python
#Newly formed classes can use the Animal class in order to inherit some of its methods that you may want to use again
```

```python
class Dog(Animal): #Inherits Animal class/Base class; now known as Derived Class 

    def __init__(self): 
        Animal.__init__(self) #Creates an instance of Animal class when I create an instance of my Dog class
        print("Dog Created")  #Able to do so because I am inheriting from the Animal class

    def who_am_i(self):
        print("I am a Dog")

    def eat(self):
        print("I am dog and eating") #Overwrote the original method 
        
    def bark(self):
        print("WOOF!")
```

```python
mydog = Dog()
```

```python
mydog.eat() #All the methods for Animal are now available for Dog, which can be overwritten 
#Inheritence: You can reuse methods common between classes by just inheriting 
```

```python
mydog.who_am_i() #You can overwrite the methods of the base class 
```

```python
myanimal = Animal()
```

```python
myanimal.eat()
```

```python
myanimal.who_am_i()
```

```python
mydog.bark()
```
