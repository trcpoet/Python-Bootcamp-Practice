### Problem 1

Create a generator that generates the squares of numbers up to some number N.

```python
def gensquares(N):
    for x in range(N):
        yield x**2
```

```python
for x in gensquares(10):
    print(x)
```

### Problem 2

Create a generator that yields "n" random numbers between a low and high number (that are inputs). <br>Note: Use the random library. For example:

```python
import random

random.randint(1,10)
```

```python
def rand_num(low,high,n):
    for i in range(n):
        yield random.randint(low,high) #yield it in order to create a generator out of the function
```

```python
for n in rand_num(1, 100, 20):#(low, high, number of results)
    print(n)
```

Can you explain what *gencomp* is in the code below? (Note: We never covered this in lecture! You will have to do some Googling/Stack Overflowing!)

```python
# The variable gencomp is a generator expression—a concise way to create a generator without the need to define a full generator function with a yield statement. Here's what happens in the code:

```

```python
my_list = [1,2,3,4,5]

gencomp = (item for item in my_list if item > 3)

for item in gencomp:
    print(item)
```
