```python
class Dog():
    
    
    #CLASS OBJECT ATTRIBUTE 
    #SAME FOR ANY INSTANCE OF A CLASS
    species = "mammal"
    

    def __init__(self, breed, name):
        # Attributes
        # We take in the argument
        # Assign it using self.attribute_name
        self.breed = breed
        self.name = name
        

    #OPERATIONS AND ACTIONS ----> METHODS
    #METHODS LOOK LIKE FUNCTIONS IN A CLASS 

    def bark(self, number): #Methods can take outside arguments - number
        print("WOOF! My name is {} and the number is {}".format(self.name, number)) 
        #number does not need self.number because number is already being provided for us when we call bark()
        #WE need to pass in the number
        
```

```python
my_dog = Dog("Lab", "Frankie")
```

```python
type(my_dog)
```

```python
my_dog.species
```

```python
my_dog.name
```

```python
my_dog.bark(5)
```

```python
class Circle():
    #CLASS OBJECT ATRIBUTE
    pi = 3.14

    def __init__(self, radius=1):
        self.radius = radius
        self.area = radius*radius*Circle.pi

    def get_circumference(self):
        return self.radius * self.pi *2
```

```python
my_circle = Circle()
```

```python
my_circle.pi
```

```python
my_circle.radius
```

```python
my_circle = Circle(30)
```

```python
my_circle.radius
```

```python
my_circle.pi
```

```python
my_circle.get_circumference()
```

```python
my_circle.area
```

```python
class Animal(): #Base Class
    def __init__(self):
        print("Animal Created")

    def who_am_i(self):
        print("I am an animal")

    def eat(self):
        print("I am eating")
```

```python
class Dog(Animal):

    def __init__(self):
        Animal.__init__(self)
        print("Dog Created")


    def eat(self):
        print("I am dog and eating")
        
    def bark(self):
        print("WOOF!")
```

```python
mydog = Dog()
```

```python
mydog.eat()
```

```python
mydog.who_am_i()
```

```python
myanimal = Animal()
```

```python
myanimal.eat()
```

```python
myanimal.who_am_i()
```

```python
mydog.bark()
```

```python
class Dog():

    def __init__(self, name):
        self.name = name

    def speak(self):
        return self.name + " says woof!"
```

```python
class Cat():

    def __init__(self, name):
        self.name = name

    def speak(self):
        return self.name + " says meow!"
```

```python
niko = Dog("niko")
felix = Cat("felix")
```

```python
print(niko.speak())
```

```python
print(felix.speak())
```

```python
for pet in [niko,felix]:
    print(type(pet))
    print(pet.speak())
```

```python
def pet_speak(pet):
    print(pet.speak())
```

```python
pet_speak(niko)
```

```python
pet_speak(felix)
```

```python
class Animal():
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclass must implement this abstract method")
```

```python
class Dog(Animal):

    def speak(self):
        return self.name +" says woof!!"
```

```python
class Cat(Animal):

    def speak(self):
        return self.name +" says meow!!"
```

```python
fred = Dog("Fred")
```

```python
simba = Cat("Simba")
```

```python
fred.speak()
```

```python
simba.speak()
```
